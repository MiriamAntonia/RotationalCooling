```python
import numpy as np
import matplotlib.pyplot as plt

with open("Jmax70_res008_shiftm100_1000_40000_2.txt", "r") as f:
    counts_last = [float(line.strip()) for line in f]

J_max=70

print(counts_last)
```

```python
# Bar positions
x = np.arange(J_max+1)
width = 0.4

# Plot
plt.figure(figsize=(8.8, 5))
#plt.bar(x - width/2, counts_first, width=width, color='lightcoral', label='Before cooling')
plt.bar(x, counts_last, width=width, color='skyblue', label='After cooling')

plt.xlabel('J', fontsize=16)
plt.ylabel('Population (%)', fontsize=16)

plt.xticks(range(0, J_max+1, 2), fontsize=14)
plt.yticks(fontsize=14)

plt.xlim([-0.5, 35.5])

plt.legend(fontsize=14)
plt.tight_layout()
plt.savefig('distribution_after_cooling.pdf')
plt.show()
```
